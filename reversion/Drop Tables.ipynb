```sql
%python
dbutils.widgets.removeAll()
```

```sql
create widget text storageName default "saejercdatabricks";
```

```sql
%python
storageName = dbutils.widgets.get("storageName")
```

###Tablas Bronze

```sql
DROP TABLE IF EXISTS cat_ejec_dbricks.bronze.suspensiones;
DROP TABLE IF EXISTS cat_ejec_dbricks.bronze.reinstalaciones;
```

```sql
%python

ruta = f"abfss://bronze@{storageName}.dfs.core.windows.net/suspensiones"

try:
    dbutils.fs.ls(ruta)
    path_exists = True
except:
    path_exists = False

if path_exists:
    dbutils.fs.rm(ruta, True)

ruta = f"abfss://bronze@{storageName}.dfs.core.windows.net/reinstalaciones"

try:
    dbutils.fs.ls(ruta)
    path_exists = True
except:
    path_exists = False

if path_exists:
    dbutils.fs.rm(ruta, True)
```
